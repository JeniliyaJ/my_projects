## Step 4: Feature Engineering (new useful columns)

```python
concrete_data['binder']= concrete_data['cement']+concrete_data['ash']+concrete_data['slag']
concrete_data['cem_wat_ratio']=concrete_data['cement']+concrete_data['water']
concrete_data['wat_bin_ratio']=concrete_data['binder']/concrete_data['water']
concrete_data['log_age']=np.log1p(concrete_data['age'])
```

## Step 5: Train / Test split (sariyaana vazhi)

```python
## intha dataset la mixid ila column same group ku same id irukanum so atha nammala create pannanum
mix_col=['cement', 'slag', 'ash', 'water', 'superplastic', 'coarseagg',
       'fineagg']
groups=concrete_data[mix_col].round(1).astype(str).agg('-'.join,axis=1)
```

```python
groups
```

```python
x=concrete_data.drop(columns=['strength'])
y=concrete_data['strength']
```

```python
splitter=GroupShuffleSplit(n_splits=1,random_state=42,test_size=0.2)
train_idx,test_idx=next(splitter.split(x,y,groups))
x_train=x.iloc[train_idx]
x_test=x.iloc[test_idx]
y_train=y.iloc[train_idx]
y_test=y.iloc[test_idx]
g_train = groups.iloc[train_idx]
```

## Data Preproccessing

```python
scaled=StandardScaler()
x_train_scaled=scaled.fit_transform(x_train)
x_test_scaled=scaled.transform(x_test)
```

## build model

```python
model={'linear_model':LinearRegression(),
       'lasso_reg':Lasso(alpha=0.1),
        'ridge_reg' :Ridge(alpha=0.01),
        'random_forest':RandomForestRegressor(n_estimators=100,max_depth=5,min_samples_split=10),
       'xboost':XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)

      }

result=[]
for name,model in model.items():
    model.fit(x_train_scaled,y_train)
    y_pred=model.predict(x_test_scaled)

    mae=mean_absolute_error(y_test,y_pred)
    mse=mean_squared_error(y_test,y_pred)
    rmse=np.sqrt(mse)
    r2=r2_score(y_test,y_pred)

    result.append([name,mae,mse,rmse,r2])
    
print(result)
```

```python
# convert the result into dataframe
```

```python
result_df=pd.DataFrame(result,columns=['model_name','mae','mse','rmse','r2']).sort_values(by='r2',ascending=False)
result_df
```

```python
## from this comparision i will choose the xboost and randomforest are best_model for my data
```

## Step 13: Best model select + test-la visualize

```python
best_model=XGBRegressor()
best_name='xboost'
best_model.fit(x_train_scaled,y_train)
y_train_pred=best_model.predict(x_train_scaled)
y_test_pred=best_model.predict(x_test_scaled)
## model evaluation
r2_test=r2_score(y_test,y_test_pred)
r2_train=r2_score(y_train,y_train_pred)
```

#### overfitting checking

```python
overfit_check=r2_train-r2_test
overfit_check
if overfit_check<0.5:
    print("no overfit")
else:
    print("overfit")
```

## Step 15: Business use 1 - What-if tool
#### mix kuduthu (cement,water) intha age 3 and 28 la ennoda concrete strength evlo irukum nu predict pannna sollanum its is use full for the site engineers

```python
concrete_data.columns
```

```python

def add_features(df):
    df = df.copy()
    df['binder'] = df['cement'] + df['slag'] + df['ash']
    df['cem_wat_ratio'] = df['cement'] / df['water']
    df['wat_bin_ratio'] = df['water'] / df['binder']
    df['log_age'] = np.log(df['age'])
    return df
```

```python
import pandas as pd

def predict_strength(cement, slag, ash, water, superplastic, coarseagg, fineagg, age):
    # 1. Base input values-ai DataFrame-ah mathrom
    row = pd.DataFrame({
        'cement': [cement], 'slag': [slag], 'ash': [ash], 'water': [water],
        'superplastic': [superplastic], 'coarseagg': [coarseagg], 'fineagg': [fineagg], 'age': [age]
    })
    
    # 2. Romba Mukkiyam: Ungaloda munnadi ulla cell-la irukra 'add_features' function-ai inga call panrom.
    # Ithuthaan antha extra columns-ai (binder, ratios, etc.) create pannum.
    row = add_features(row)
    
    # 3. Model predict panrathu
    return float(best_model.predict(row)[0])

# Example test:
for d in [3, 7, 28, 90]:
    s = predict_strength(300, 100, 0, 180, 6, 950, 780, d)
    print(f"Age {d:3d} days -> {s:.1f} MPa")
```

## Step 16: Business use 2 - Cheapest mix finder

```python
import pandas as pd
import numpy as np
from scipy.optimize import minimize


PRICE = {
    'cement': 0.12,
    'slag': 0.05,
    'ash': 0.03,
    'water': 0.001,
    'superplastic': 1.5,
    'coarseagg': 0.02,
    'fineagg': 0.02
}

def find_cheapest_mix(target_strength=40.0, age=28):
    # Total cost-ai calculate panra function
    def calculate_cost(mix):
        return (
            mix[0] * PRICE['cement'] +
            mix[1] * PRICE['slag'] +
            mix[2] * PRICE['ash'] +
            mix[3] * PRICE['water'] +
            mix[4] * PRICE['superplastic'] +
            mix[5] * PRICE['coarseagg'] +
            mix[6] * PRICE['fineagg']
        )

    # Strength constraint function (add_features use panni 12 columns-ai create panrom)
    def strength_constraint(mix):
        row = pd.DataFrame([[
            mix[0], mix[1], mix[2], mix[3], 
            mix[4], mix[5], mix[6], age
        ]], columns=['cement', 'slag', 'ash', 'water', 'superplastic', 'coarseagg', 'fineagg', 'age'])
        
        # Romba mukkiyam: 12 columns-ai generate panrathukku add_features-ai call panrom
        row = add_features(row)
        
        predicted = float(best_model.predict(row)[0])
        # Target strength-ku mela irukanum
        return predicted - target_strength

    # Ovvoru ingredient-kku min & max limits (Bounds)
    bounds = [
        (100, 500),  # cement
        (0, 200),    # slag
        (0, 150),    # ash
        (100, 250),  # water
        (0, 20),     # superplastic
        (800, 1100), # coarseagg
        (600, 900)   # fineagg
    ]

    # Oru starting guess alavu
    initial_guess = [300, 50, 20, 180, 5, 950, 750]

    # Optimization run panrom
    constraints = {'type': 'ineq', 'fun': strength_constraint}
    result = minimize(calculate_cost, initial_guess, method='SLSQP', bounds=bounds, constraints=constraints)

    if result.success:
        print(f"🎯 Cheapest Mix Found for Target Strength: {target_strength} MPa (Age: {age} days)")
        print(f"💰 Minimum Cost: ${result.fun:.2f}")
        opt = result.x
        print(f"   - Cement: {opt[0]:.1f} kg")
        print(f"   - Slag: {opt[1]:.1f} kg")
        print(f"   - Fly Ash: {opt[2]:.1f} kg")
        print(f"   - Water: {opt[3]:.1f} kg")
        print(f"   - Superplasticizer: {opt[4]:.1f} kg")
        print(f"   - Coarse Aggregate: {opt[5]:.1f} kg")
        print(f"   - Fine Aggregate: {opt[6]:.1f} kg")
        return opt
    else:
        print("Optimization failed. Bounds-ai check pannunga.")
        return None

best_mix = find_cheapest_mix(target_strength=40.0, age=28)
```
